```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd
import plotly.express as px
from src.analysis import (
    load_processed, filter_data, kpis, serie_temporal_nacional,
    agregacao_uf, correlacao_agua_saude, casos_por_faixa_etaria,
    ranking_municipios,
)

df = load_processed()
print(df.shape)
df.head()
```

```python
# Recorte padrão do dashboard (2024–2026) vs série completa
full_kpis = kpis(df)
recent = filter_data(df, ano_min=2024, ano_max=2026)
print("Completo:", full_kpis)
print("2024–2026:", kpis(recent))
```

```python
serie = serie_temporal_nacional(df)
fig = px.line(serie, x="data_ref", y=["incidencia_10k", "taxa_ecoli"],
              title="Incidência DDA /10k e taxa E. coli (nacional)")
fig.show()
```

```python
uf = agregacao_uf(df)
px.scatter(uf, x="taxa_ecoli", y="incidencia_10k", size="n_municipios",
           hover_name="uf", title="UF: E. coli × incidência DDA").show()
correlacao_agua_saude(df)
```
